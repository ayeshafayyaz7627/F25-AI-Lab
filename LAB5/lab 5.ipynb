```python
import math
count_mm = 0
def minimax(depth, index, maxTurn, scores, target):
    global count_mm
    if depth == target:
        count_mm += 1
        return scores[index]
    left = minimax(depth+1, index*2, not maxTurn, scores, target)
    right = minimax(depth+1, index*2+1, not maxTurn, scores, target)
    return max(left, right) if maxTurn else min(left, right)
count_ab = 0
def alphabeta(depth, index, maxTurn, scores, target, alpha, beta):
    global count_ab
    if depth == target:
        count_ab += 1
        return scores[index]
    if maxTurn:
        value = float('-inf')
        for i in range(2):
            value = max(value, alphabeta(depth+1, index*2+i, False, scores, target, alpha, beta))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for i in range(2):
            value = min(value, alphabeta(depth+1, index*2+i, True, scores, target, alpha, beta))
            beta = min(beta, value)
            if alpha >= beta:
                break
        return value
def run(scores):
    global count_mm, count_ab
    count_mm = count_ab = 0
    d = int(math.log2(len(scores)))
    r1 = minimax(0, 0, True, scores, d)
    r2 = alphabeta(0, 0, True, scores, d, float('-inf'), float('inf'))
    print("Scores:", scores)
    print("Minimax value:", r1, "| leaves:", count_mm)
    print("AlphaBeta value:", r2, "| leaves:", count_ab)
    print("Same answer?", r1 == r2)
    print()
run([3, 5, 2, 9, 0, 7, 0, 0])
run([3, 5, 2, 9, 3, 5, 2, 9])
run([3, 5, 2, 9, 0, 0, 7, 0])
run([5, 3, 9, 2, 7, 0, 0, 0])
```
